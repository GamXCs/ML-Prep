```python
"""
mean
median
minimum
maximum
standard deviation

"""

names = np.array([
    "Alice",
    "Bob",
    "Charlie",
    "Dana",
    "Ethan",
    "Faith"
])

exam1 = np.array([88, 64, 91, 72, 59, 84])

exam2 = np.array([92, 70, 89, 68, 61, 95])

print(exam1.min())
print(exam1.max())
print(exam1.std())
print(exam1.mean())

passed_exam1 = exam1 >= 70
print(exam1[passed_exam1])

passed_exam2 = exam2 >= 70
print(exam2[passed_exam2])

# print names of students who passed both exams
passed_both = passed_exam1 & passed_exam2
print(names[passed_both])
```

```python
# print averages for each student
exams_averages = (exam1 + exam2) / 2
print(exams_averages)
```

```python
# find best performer
highest_score = exams_averages.argmax()
print(names[highest_score])
```

```python
# find the students whose individual avg exceeds overall student avg

avg = exams_averages.mean()
above_mask = exams_averages > avg
print(names[above_mask])
```

```python
# In ML lang: (num of samples, num of features)

X = np.column_stack((exam1, exam2))
print(X.shape)
```

```python
# cols: Study Hours and Previous Exam
names = np.array([
    "Alice",
    "Bob",
    "Charlie",
    "Dana",
    "Ethan"
])

X = np.array([
    [4,65],
    [7,78],
    [9,92],
    [3,70],
    [8,88]
])

# weights
w = np.array([
    2.5,
    0.7
])

b = 5
```

```python
# print shape, dimensions, size
X.shape
```

```python
X.ndim
```

```python
X.size
```

```python
"""
Using only axis, compute:
Mean study hours
Mean previous exam
Minimum of each feature
Maximum of each feature
Standard deviation of each feature
"""
study_hours_mean = X[:,0].mean(axis=0)
mean_prev_exam = X[:,1].mean(axis=0)
min_of_study_hrs = X[:,0].min()
min_of_prev_exam = X[:,1].min()
max_of_study_hrs = X[:,0].max()
max_of_prev_exam = X[:,1].max()
print(f"Previous exam mean: {mean_prev_exam}")
print(f"Study hours mean: {study_hours_mean}")
print(f"Min of study hours: {min_of_study_hrs}")
print(f"Min of previous exam: {min_of_prev_exam}")
print(f"Max of study hours: {max_of_study_hrs}")
print(f"Max of previous exam: {max_of_prev_exam}")
```

```python
# compute predictions with formula: X @ + b 
# find highest and lowest prediction

predictions = X @ w + b
highest_prediction = predictions.max()
lowest_prediction = predictions.min()
print(f"Predictions: {predictions}")
print(f"Highest prediction: {highest_prediction}")
print(f"Lowest prediction: {lowest_prediction}")
```

```python
# find predictions greater than or equal to 75
# print names and predictions of those students
# find the student with the highest prediction

pred_over_75 = predictions >= 75
highest_idx = predictions.argmax()
print(f"Predictions over 75: {predictions[pred_over_75]}")
print(f"Names of predictions over 75: {names[pred_over_75]}")
print(f"Highest Prediction: {names[highest_idx]} {highest_prediction:.2f}")
```

```python
# compute the norm of each students feature vector

norms = np.linalg.norm(X,axis=1)
for name, norm in zip(names, norms):
    print(f"{name} {norm:.2f}")
```
