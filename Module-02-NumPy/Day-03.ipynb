```python
import numpy as np
```

```python
student_names = np.array(["Amina", "Carlos", "Jade", "Malik"])

X = np.array([
    [8, 6, 9],
    [5, 7, 6],
    [9, 8, 8],
    [6, 5, 7]
], dtype=float)

w = np.array([0.4, 0.35, 0.25])
b = 2.0
```

```python
# calculate predictions
predictions = X @ w + b
```

Attach students with the prediction scores. Doing this does not create a new array, but, the two arrays are treated as parallel. This means that we can use argmax to get location of elements in student


```python
for student, pred in zip(student_names, predictions):
    print(f"{student}: {pred:.2f}")
```

```python
# print highest scoring student

highest_stu_idx = predictions.argmax()
highest_student = student_names[highest_stu_idx]
highest_prediction = predictions[highest_stu_idx]
print(f"{highest_student}: {highest_prediction:.2f}")
```

```python
# calculate mean of predictions and id students above the mean

pred_mean = predictions.mean()
above_pred_mean_mask = predictions > pred_mean
above_pred_mean_students = student_names[above_pred_mean_mask]
above_pred_mean_student_scores = predictions[above_pred_mean_mask]

# join names and scores and display
for stu, score in zip(above_pred_mean_students, above_pred_mean_student_scores):
    print(f"{stu}: {score}")
```

```python
# turn prediction calculation into a function

def prediction_func(X, w, b):
    return X @ w + b

# use np.allclose to verify func predictions are same as manaully calc pred

np.allclose(prediction_func(X, w, b), predictions)
```
