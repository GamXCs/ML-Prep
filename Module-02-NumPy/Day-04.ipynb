```python
names = np.array([
    "Amina",
    "Carlos",
    "Jade",
    "Malik"
])

X = np.array([
    [8, 6],
    [5, 7],
    [9, 8],
    [3, 4]
], dtype=float)

new_student = np.array([7, 7])
```

- Column 0 → hours studied
- Column 1 → practice assignments completed


Which existing student is most similar to the new student?


```python
students_differences = X - new_student
students_differences
# students_differences will give a 4,2 matrix showing the distance of 
# each student from new_student
```

```python
distances = np.linalg.norm(students_differences, axis=0)
distances
```

```python
euclid_dist = np.linalg.norm(students_differences, axis=1)
euclid_dist
```

```python
for name, distance in zip(names, euclid_dist):
    print(f"{name}: {distance:.2f}")
```

```python
# get student closest to new_student

closest_idx = euclid_dist.argmin()
closest_student= names[closest_idx]
closest_student_distance = euclid_dist[closest_idx]
print(f"Nearest student: {closest_student}")
print(f"Distance: {closest_student_distance:.2f}")
```

The above was demonstrating k-NN (K-Nearest Neighbors) w/o sklearn


```python
# add labels

passed = np.array([
    True,   # Amina
    True,   # Carlos
    True,   # Jade
    False   # Malik
])
```

```python
# get predict_pass value from passed matrix

predicted_pass = passed[closest_idx]
predicted_pass
```

```python
# turn it into a classifier function

# new_observation is like new_student from earlier example

def nearest_neighbor_predict(X, y, new_observation):
    """X: feature matrix containing existing observations
       y: target labels corresponding to said observations
       new_observation: features of the observation we're predicting
    """
    differences = X - new_observation
    X_distance = np.linalg.norm(differences, axis=1)
    nearest = X_distance.argmin()
    predicted_label = y[nearest]
    return predicted_label
    
nearest_neighbor_predict(X, passed, new_student)
test_student = np.array([3, 3])
nearest_neighbor_predict(X, passed, test_student)
```

```python
assert nearest_neighbor_predict(X, passed, test_student) == False
assert nearest_neighbor_predict(X, passed, new_student) == True
print("All tests passed")
```
