```python
# now we see if the model can predict things it has never seen

Y_predictions = model.predict(X_test_scaled)
```

```python
# compare predictions and labels
y_pred = Y_predictions
y_lab = Y_test

print(f"Y prediction values: {y_pred}")
print(f"Y label values: {y_lab}")

""" y_pred are the values the machine generated after we fed it the scaled data.
    y_test are the actual label (true data) that came with original dataset 
"""
```

```python
# calculate the accuracy : num correct pred/ total pred

correct_pred = (y_pred == y_lab).sum()
total_pred = len(y_pred)
accuracy = correct_pred/total_pred
accuracy_per = accuracy * 100
print(f"Accuracy: {accuracy_per:.2f}%")
```

```python
# create confusion matrix to investigate the mistakes

confusion_matx = confusion_matrix(y_lab, y_pred)
confusion_matx
```

### Interpretation

- **TN = 71:** The model correctly identified 71 benign tumors.

- **FN = 1:** The model incorrectly classified one malignant tumor as benign.

### Calculate Recall for malignant tumors

$$
\text{Recall} = \frac{TP}{TP + FN}
$$

```python
TP = confusion_matx[0, 0]
TN = confusion_matx[1, 1]
FP = confusion_matx[1, 0]
FN = confusion_matx[0, 1]
```

```python
recall = (TP / (TP + FN)) * 100
print(f"Recall: {recall:.2f}%")
print(f"Of the 42 malignant tumors, {recall:.2f}% were correctly identified")
```

### Calculate Precision for Malignant Tumors

$$
\text {Precision} = \frac{TP}{TP + FP}
$$

```python
precision = (TP / (TP + FP)) * 100
print(f"Precision: {precision:.2f}%")
print("Out of all the tumors we predicted were malignant, approx 97% actually were")
```

### Calculate F1 Score

$$
\text{F1} = 2 \cdot \frac{precision \cdot recall}{precision + recall}
$$

```python
F1 = 2 * ((precision * recall) / (precision + recall))
print(f"F1 score: {F1:.2f}%")
```

### Realized that I would be reusing variables/behavior, so refactoring for reusability by making functions


```python
def evaluate_model(y_true, y_pred):
    """ Convention for this model is positive class is malignant(0), neg class is benign(1) """
    
    # generate confusion matrix
    confusion_mtrx = confusion_matrix(y_true, y_pred, labels=[0,1])

    # extract TP, TN, FP, FN 
    TP = confusion_mtrx[0, 0]
    TN = confusion_mtrx[1, 1]
    FP = confusion_mtrx[1, 0]
    FN = confusion_mtrx[0, 1]

    # calculate eval metrics
    # make sure to account for dividing by zero
    # standard convention is to use 0.0 
    
    if TP + FN == 0:
        recall = 0.0
    else:
        recall = (TP / (TP + FN)) * 100
    
    if TP + FP == 0:
        precision = 0.0
    else:
        precision = (TP / (TP + FP)) * 100

    if precision + recall == 0:
        F1 = 0.0
    else:    
        F1 = 2 * ((precision * recall) / (precision + recall))
        
    accuracy = ((TP + TN) / (TP + TN + FP + FN)) * 100

    # display results
    return {
        "recall": recall,
        "precision": precision,
        "f1": F1,
        "accuracy": accuracy
    }

original_results = evaluate_model(Y_test, Y_predictions)
threshold_results = evaluate_model(Y_test, new_threshold_pred)

print("Original:", original_results)
print()
print("Threshold 0.60:", threshold_results)
```

```python
# test different thresholds to see if stats improve

thresholds = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9]

# turn benign column into variable
benign_pred = Y_probabilities[:,1]

# initialize precision and recall scores empty list
precision_scores = []
recall_scores = []


# loop through thresholds
for threshold in thresholds:
    new_thresh_pred = (benign_pred >= threshold).astype(int)

    # call evaluate_model()
    thresh_results = evaluate_model(Y_test, new_thresh_pred)

    # add precision and recall scores to the lists
    precision_scores.append(thresh_results["precision"])
    recall_scores.append(thresh_results["recall"])

    # print results
    print(f"Threshold {threshold}, {thresh_results}\n")
print(f"Precision scores: {precision_scores}\n")
print(f"Recall scores: {recall_scores}")
```

```python
# plot the data

plt.plot(thresholds, precision_scores, label="Precision")
```

```python
plt.plot(thresholds, recall_scores, label="Recall")
```

```python
# plot precision and recall against one another

plt.title("Precision and Recall vs. Classification Threshold")
plt.plot(thresholds, precision_scores, label="Precision")
plt.plot(thresholds, recall_scores, label="Recall")
plt.ylabel("Score (%)")
plt.xlabel("Classification Threshold")
plt.legend()
plt.grid(True)
plt.show()
```
